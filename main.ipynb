## Pandas Basics

### Importing Pandas

The Pandas library must always be imported on top of your script.

```python
import pandas as pd
```

```python
# Helper function to create a DataFrame
def create_dataframe():
    
    data = {
        'Name': ['Alice', 'Bob', 'Charlie'],
        'Age': [25, 30, 35],
        'City': ['New York', 'Los Angeles', 'Chicago']
    }

    return pd.DataFrame(data)
```

### Series

A Series is like a column in a table, a one-dimensional array used to hold data of any type.

```python
# Create a Series with a list of numbers
aList = [1, 2, 3, 4, 5]

numbers = pd.Series(aList)

# Display the Series
print(aList)
```

It can also hold key/value pairs, like a dictionary.

```python
# Create a Series with key-value pairs
aDict = {'a': 1, 'b': 2, 'c': 3, 'd': 4, 'e': 5}

numbersDict = pd.Series(aDict)

# Display the Series    
print(numbersDict)
```

### DataFrames

A DataFrame is a two-dimensional data structure, AKA a table with rows and columns.

```python
# Create a DataFrame with some sample data
df = create_dataframe()

df
```

Take note that the row index is displayed to the left of the DataFrame. Row indexes in Pandas start with 0 by default.

## Pandas Properties and Methods

### `loc` Property

`loc` returns a specific row in your DataFrame, this will return a Series

```python
# Create a DataFrame with some sample data
df = create_dataframe()

df.loc[0]
```

You can also return multiple rows by passing a list, this will return a DataFrame

```python
df.loc[[0, 1]] 
```

A table from a Wikipedia, Pandas reads in the raw HTML and indexes the `<table>...</table>` tags

```python
# Inspect the code from the URL below
url = 'https://en.wikipedia.org/wiki/List_of_Formula_One_World_Drivers%27_Champions'

# Read HTML from the specified URL
import requests
html = requests.get(url).text

print(html) # Display the HTML content retrieved from the URL
```

```python
# Read all tables from the specified URL
df = pd.read_html(url)

# Loop through the tables and display the first 5 rows of each table
for i, table in enumerate(df):

    print(f"Table {i}:")

    print(table.head())  # Display the first 5 rows of each table
    
    print("\n")
```

### Column Operations

```python
# Create a DataFrame with some sample data
df = create_dataframe()

columns = ['Name', 'Age']

df[columns]  # Display only the specified columns
```

```python
# Change the name of a column
df.rename(columns={'Name': 'Full Name'}, inplace=True) # inplace=True modifies the DataFrame directly
df
```

```python
# Add a new column to the DataFrame
df['Country'] = ['USA', 'USA', 'USA']
df
```

```python
# Subtract 5 from the 'Age' column
df['Age'] = df['Age'] - 5

# df['Age'] -= 5  # Another way to subtract 5 from the 'Age' column

df
```

```python
# Remove the Country column
df.drop(columns=['Country'], inplace=True)
df
```

```python
# Change the name on the first row to 'Alicia' 
df.loc[0, 'Full Name'] = 'Alicia'
df
```

### Filtering by Value

```python
# Create a DataFrame with some sample data
df = create_dataframe()

df[df['Age'] >= 30]  # Filter rows where Age is greater than or equal to 30
```

```python
# Get a list of names in the DataFrame where the City contains the letter 'e' and print out the names with a message 

letter = 'e'

names_with_letter = df[df['City'].str.contains(letter)]['Name']

for name in names_with_letter:
    
    print(f"{name} lives in a city that contains the letter '{letter}'.")
```

### iterrows()

```python
# Create a DataFrame with some sample data and iterate through its rows

df = create_dataframe()

template = 'Row {ix}: {name} is {age} years old and lives in {city}.'

for index, row in df.iterrows():

    print(template.format(ix=index, name=row['Name'], age=row['Age'], city=row['City']))
```
